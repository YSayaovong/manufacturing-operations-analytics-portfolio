# Engineering change order workflow analysis
Synthetic data. Business question: which change types are associated with longer approval cycles and greater production disruption?

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('https://raw.githubusercontent.com/YSayaovong/manufacturing-operations-analytics-portfolio/main/06-engineering-changes/data.csv', parse_dates=['submitted_date','approved_date'])
assert (df.approval_days == (df.approved_date-df.submitted_date).dt.days).all()
print(df[['approval_days','production_delay_days','estimated_delay_cost_usd']].describe())
```

```python
result = df.groupby('change_type').agg(changes=('eco_id','count'),mean_approval_days=('approval_days','mean'),total_delay_days=('production_delay_days','sum'),total_estimated_cost=('estimated_delay_cost_usd','sum')).sort_values('total_estimated_cost',ascending=False)
print(result.round(1))
result.total_estimated_cost.plot.bar(title='Estimated delay cost by ECO type',ylabel='Synthetic USD')
plt.tight_layout();plt.show()
```

## Action plan
Prioritize approval bottlenecks using mean and median lead time; segment by change type and team. Estimate improvement potential with explicit assumptions. Association is not proof that an ECO caused the delay.
